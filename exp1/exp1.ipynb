```python
# importing libraries
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
# reading image
img = cv2.imread("/Users/vanshilpatel/Desktop/cv pdeu/exp2/Image.png")
img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
if img is None:
    print("Image not found!")
    exit()
```

```python
# displaying the image
print("Original Image")
plt.axis('off')
plt.imshow(img)
plt.title("Original")
plt.show()
```

```python
# getting image shape and data type
print("Image Shape :", img.shape)
print("Image Data Type :", img.dtype)
```

```python
# cropping the image
y_start, y_end = 20, 120
x_start, x_end = 50, 150

cropped_img = img[y_start:y_end, x_start:x_end]

print("Original Shape :", img.shape)
print("Cropped Shape :", cropped_img.shape)

plt.imshow(cropped_img)
plt.title("Cropped Image")
plt.axis("off")
plt.show()
```

```python
# resizing the image
resized = cv2.resize(img, (300, 500))

plt.imshow(resized)
plt.title("Resized Image")
plt.axis("off")
plt.show()
```

```python
# changing a pixel value
modified = img.copy()

modified[100, 100] = [0, 0, 255]

plt.imshow(modified)
plt.title("Modified Image")
plt.axis("off")
plt.show()
```

```python
# converting to grayscale
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

plt.imshow(gray, cmap='gray')
plt.title("Grayscale Image")
plt.axis("off")
plt.show()
```

```python
# converting to RGB
rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

plt.imshow(rgb)
plt.title("RGB Image")
plt.axis("off")
plt.show()
```

```python
# flipping the image horizontally
flipped = cv2.flip(img, 1)

plt.imshow(flipped)
plt.title("Flipped Image")
plt.axis("off")
plt.show()
```

```python
# rotating the image 90 degrees clockwise
rotated = cv2.rotate(img, cv2.ROTATE_90_CLOCKWISE)

plt.imshow(rotated)
plt.title("Rotated Image")
plt.axis("off")
plt.show()
```

```python
# applying Gaussian blur
blur = cv2.GaussianBlur(img, (13, 13), 0)

plt.imshow(blur)
plt.title("Gaussian Blur Image")
plt.axis("off")
plt.show()
```
