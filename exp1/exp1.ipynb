```python
# importing libraries
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
# reading image
img = cv2.imread("/Users/vanshilpatel/Desktop/cv pdeu/exp1/Image.png")
if img is None:
    print("Image not found!")
    exit()
```

```python
# displaying the image
print("Original Image")
plt.imshow(img)
plt.title("Original")
plt.show()
```

```python
# getting image shape and data type
print("Image Shape :", img.shape)
print("Image Data Type :", img.dtype)
```

```python
# cropping the image
y_start, y_end = 20, 120
x_start, x_end = 50, 150

cropped_img = img[y_start:y_end, x_start:x_end]

print("Original Shape :", img.shape)
print("Cropped Shape :", cropped_img.shape)

cv2.imshow("Cropped", cropped_img)
```

```python
# resizing the image
resized = cv2.resize(img, (300, 500))

print("\nResized Image")
cv2.imshow("Resized", resized)
```

```python
# changing a pixel value
modified = img.copy()

modified[100, 100] = [0, 0, 255]

print("\nOne Pixel Changed")
cv2.imshow("Modified", modified)
```

```python
# converting to grayscale
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

print("\nGrayscale Image")
cv2.imshow("Gray", gray)
```

```python
# converting to RGB
rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

plt.imshow(rgb)
plt.title("RGB Image")
plt.axis("off")
plt.show()
```

```python
# flipping the image horizontally
flipped = cv2.flip(img, 1)

print("\nHorizontally Flipped Image")
cv2.imshow("Flipped", flipped)
```

```python
# rotating the image 90 degrees clockwise
rotated = cv2.rotate(img, cv2.ROTATE_90_CLOCKWISE)

print("\nRotated Image (90 Degrees)")
cv2.imshow("Rotated", rotated)
```

```python
# applying Gaussian blur
blur = cv2.GaussianBlur(img, (13, 13), 0)

print("\nGaussian Blur Image")
cv2.imshow("Blur", blur)
```
